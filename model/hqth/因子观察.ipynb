```python

target_cols = [
    "fac1004", "fac1133", "fac1160", "fac1170", "fac0279", "fac1136",
    "fac1161", "fac1005", "fac1135", "fac1137", "fac1150", "fac1006",
    "fac0230", "fac1162", "fac0235", "fac0281", "fac1007", "fac0022",
    "fac0038", "fac0039", "fac0046", "fac0233", "fac0321", "fac1008",
    "fac1009", "fac1010", "fac1793", "fac2372", "fac2373", "fac2374",
    "fac2565", "fac2566"
]
target_cols += ["Code", "date"]

target_data = data[target_cols]
```

```python

import pandas as pd

# 取出第一行（可以用 .head(1)）
target_data1000 = target_data.head(1000)

# 保存为 Excel
with pd.ExcelWriter("target_data1000.xlsx") as writer:
    target_data1000.to_excel(writer, sheet_name="target_data1000", index=False)
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# 选择数据子集
target_data = data[target_cols]

# 仅保留数值列（排除Code和date）
numeric_cols = target_data.select_dtypes(include='number').columns

# 逐个绘制直方图
for col in numeric_cols:
    plt.figure(figsize=(6, 4))
    plt.hist(target_data[col].dropna(), bins=30, edgecolor='black')
    plt.title(f"Histogram of {col}")
    plt.xlabel(col)
    plt.ylabel("Frequency")
    plt.grid(True)
    plt.tight_layout()
    plt.show()
```

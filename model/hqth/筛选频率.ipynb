```python
import pandas as pd 
import re


data_list = ["2023q1_num-1400.fea", "2023q2_num-1400.fea", "2023q3_num-1400.fea", "2023q4_num-1400.fea", "2024q1_num-1400.fea", "2024q2_num-1400.fea", "2024q3_num-1400.fea", "2024q4_num-1400.fea"]

# 读取特征列
with open('0708categorical_features.txt', 'r') as f:
    categorical = [line.strip() for line in f.readlines()]

with open('0708continuous_features.txt', 'r') as f:
    continuous = [line.strip() for line in f.readlines()]


commons = []
for data_name in data_list:
    # 读取 Feather 文件
    data = pd.read_feather(rf'/project/model_share/share_1_new/factor_csy/fac20250613_ic1400/{data_name}')
    print("✅ Feather 文件读取完成！")
    # 提取所有 fac 列名
    fac_cols = [col for col in data.columns if col.startswith('fac')]

    common = [elem in fac_cols for elem in categorical]
    commons.append(common)
    print(f"{data_name} 中共有 {sum(common)} 个离散数据, 占比 {sum(common)/len(fac_cols):.2%}。")
```

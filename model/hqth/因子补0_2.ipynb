```python
import pandas as pd 


# 读取 Feather 文件
data = pd.read_feather(rf'/home/user96/label_data1/label1.fea')

print("✅ Feather 文件读取完成！")
```

```python
data.info
```

```python

# 获取所有唯一的日期和代码
all_dates = data['date'].unique()
all_codes = data['Code'].unique()

# 创建所有 (date, Code) 的组合
full_index = pd.MultiIndex.from_product([all_dates, all_codes], names=['date', 'Code'])
full_df = pd.DataFrame(index=full_index).reset_index()

# 合并原始数据
merged = pd.merge(full_df, data, on=['date', 'Code'], how='left')

# 用 0 填充缺失值
merged.fillna(0, inplace=True)
```

```python
# 1. 重置索引，把 "date" 和 "Code" 变回普通列
merged = merged.reset_index(drop=True)

# 2. 保存为 feather 格式（.fea 只是简称，实际上是 .feather 文件）
merged.to_feather("/project/model_share/share_1_new/factor_csy/fac20250613/fill_zero.fea")
```

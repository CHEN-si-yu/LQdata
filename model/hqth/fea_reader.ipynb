```python
import pandas as pd 


# 读取 Feather 文件
data = pd.read_feather(rf'/project/model_share/share_1_new/factor_csy/fac20250708/2024q2_num-1400.fea')

print("✅ Feather 文件读取完成！")
```

```python
import pandas as pd
import re

# 提取所有 fac 列名
fac_cols = [col for col in data.columns if col.startswith('fac')]

# 提取 fac 后的数字并排序
fac_cols_sorted = sorted(fac_cols, key=lambda x: int(re.findall(r'\d+', x)[0]))
```

```python
# 创建 DataFrame 存储列名
fac_df = pd.DataFrame({'fac_columns_sorted': fac_cols_sorted})

# 保存为 Excel
fac_df.to_excel('sorted_fac_columns.xlsx', index=False)
```

```python
import os

directory = '/project/zjh/zjh_remote/fac_0701'
files = os.listdir(directory)
print(files)
```

```python
import os
import pandas as pd
from tqdm import tqdm  # 进度条库

datas = []

# 使用 tqdm 包装 files 列表以显示进度
for file_name in tqdm(files, desc="读取文件进度"):
    date = file_name.split(".")[0]  # 提取日期
    if date > "20241231":
        continue
    data = pd.read_feather(os.path.join(directory, file_name))
    data['date'] = date  # 添加日期列
    datas.append(data)

# 合并所有 DataFrame
merged_data = pd.concat(datas, ignore_index=True)
```

```python
merged_data.sort_values(by=['date','Code'], inplace=True)
merged_data.set_index("date", inplace=True)
merged_data.reset_index(inplace=True)
# merged_data.drop(columns=['index'], inplace=True)
# merged_data.reset_index(drop=True, inplace=True)
```

```python
import pandas as pd
import numpy as np

np.random.seed(42)  # 保证结果可复现

# 参数
n_dates = 10
n_codes_per_date = 100
n_factors = 20

dates = pd.date_range("2023-01-01", periods=n_dates, freq="W")
codes = [f"code{i:04d}" for i in range(n_codes_per_date)]

# 构建date索引
index = pd.MultiIndex.from_product([dates, codes], names=['date', 'Code'])

# 创建随机数据并插入缺失值
data_array = np.random.randn(len(index), n_factors).astype(np.float32)
mask = np.random.rand(*data_array.shape) < 0.1  # 10% NaN
data_array[mask] = np.nan

# 构造 DataFrame
demo_data = pd.DataFrame(data_array, index=index, columns=[f"factor_{i}" for i in range(n_factors)])

# 按照date作为索引
demo_data = demo_data.reset_index().set_index('date')
```

```python
for date, group in demo_data.groupby(level='date'):
    a = date
    b = group
    b = b.reset_index()
    b = b.drop(columns=['date', 'Code'])
    break
```

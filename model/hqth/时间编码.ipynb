```python
import numpy as np
import pandas as pd
import os


data = pd.read_feather("/home/user96/score_file/barra.fea")
data
```

```python
# 1. 将 date 转换为 datetime 类型
data["date"] = pd.to_datetime(data["date"].astype(str), format="%Y%m%d")

# 2. 提取年份
data["year"] = data["date"].dt.year

# 3. 提取季度、月份、周
data["quarter"] = data["date"].dt.quarter
data["month"] = data["date"].dt.month
data["week"] = data["date"].dt.isocalendar().week.astype(int)

# 4. 用 sin 和 cos 做周期特征编码
# 月份 (12个月周期)
data["month_sin"] = np.sin(2 * np.pi * data["month"] / 12)
data["month_cos"] = np.cos(2 * np.pi * data["month"] / 12)

# 季度 (4个季度周期)
data["quarter_sin"] = np.sin(2 * np.pi * data["quarter"] / 4)
data["quarter_cos"] = np.cos(2 * np.pi * data["quarter"] / 4)

# 周 (52周周期)
data["week_sin"] = np.sin(2 * np.pi * data["week"] / 52)
data["week_cos"] = np.cos(2 * np.pi * data["week"] / 52)

data.drop(columns=["quarter", "month", "week"], inplace=True)

# 将 date 转回 8 位字符串
data["date"] = data["date"].dt.strftime("%Y%m%d")
data = data[["date", "Code", "year", "month_sin", "month_cos", "quarter_sin", "quarter_cos", "week_sin", "week_cos"]]

data.to_feather(f"/home/user96/score_file/time.fea")

data
```

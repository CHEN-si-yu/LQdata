```python
import pandas as pd
import os
from tqdm import tqdm   # 导入 tqdm

fac_path = rf'/project/model_share/user96_factor/'
fac_name = rf'fac2025212_2017'

dir_list = os.listdir(fac_path + fac_name)

dir_save = fac_path + fac_name + "_sample"
if not os.path.exists(dir_save):
    os.makedirs(dir_save)

for dir in tqdm(dir_list, desc="Processing files"):
    dir_fac = fac_path + fac_name + '/' + dir
    data = pd.read_feather(dir_fac)
    other_cols = [c for c in data.columns if c not in ["Code", "date"]]
    other_cols_to_keep = other_cols[:50]
    final_cols = ["Code", "date"] + other_cols_to_keep
    data_sample = data[final_cols]
    data_sample.to_feather(dir_save + '/' + dir)
    print(f"Processed {dir}, saved to {dir_save}/{dir}")

```

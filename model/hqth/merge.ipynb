```python
import pandas as pd 
data1 = pd.read_feather("/home/user96/score_file/barra.fea")
data2 = pd.read_feather("/home/user96/score_file/hist_time_holiday.fea")
```

```python
# 只保留列名字带有holiday的列
data2 = data2[[col for col in data2.columns if 'holiday' in col or col in ['Code', 'date']]]
data2
```

```python
data2.to_feather("/home/user96/score_file/holiday.fea")
```

```python
# 然后用 merge 按照 date 和 Code 进行内连接（inner join，取交集）
df_merged = pd.merge(data2, data1, on=["date", "Code"], how="inner")
df_merged.to_feather("/home/user96/score_file/hist_time_barra.fea")
df_merged
```
